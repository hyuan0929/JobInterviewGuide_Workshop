# Question 1
**Which statement best describes supervised vs. unsupervised learning?**  
A. Supervised learning uses labeled data, while unsupervised learning does not use target labels.  
B. Supervised learning only works for regression.  
C. Unsupervised learning always performs better in high dimensions.  
D. Supervised learning requires cross-validation.  
E. Unsupervised learning cannot do dimensionality reduction.  

Your Answer: **A**  
Correct Answer: **A** ✅  

**Explanation:**  
Supervised learning uses labeled targets (X, y). Unsupervised learning works without labels (e.g., clustering, PCA).
