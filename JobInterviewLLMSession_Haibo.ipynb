1️⃣ Supervised vs Unsupervised

Quick start — how would you explain supervised vs unsupervised learning?

Nice 👍 that’s correct.

2️⃣ Linear vs Logistic

When would you use linear regression instead of logistic regression?

Linear regression predicts categories, logistic predicts values.

Hmm — actually the opposite.
Linear predicts continuous values, logistic predicts probability of classes.

3️⃣ Features vs Target

Can you define independent vs dependent variables in ML?

Independent variables are inputs, dependent is output.

Correct — though in interviews we usually say features (X) and target (y).

4️⃣ Train / Validation / Test

Why do we split data into train, validation, and test?

Train to learn, validation to tune, test for final evaluation.

Yes — good.

5️⃣ Stratification

Do you know what stratified splitting helps with?

It just makes the data more random.

Not quite — it preserves class balance across splits.

6️⃣ Data Leakage

What is data leakage?

When the model sees test data during training.

Right — can you give an example?

Maybe when using scaling?

Close — scaling before splitting is a classic leakage case.

7️⃣ Scaling Workflow

So is it okay to scale before splitting?

Yes, because scaling is just normalization.

Actually that leaks validation/test information —
we should fit scaler on training only.

8️⃣ Metrics

Candidate:

Not always — log-loss captures prediction confidence.

9️⃣ Cross-Entropy

What does log-loss penalize?

Partially — it penalizes confident wrong predictions heavily.

🔟 Logistic Coefficients

## 🪞 Reflection 

During the interview simulation, I realized that my practical coding experience is stronger than my ability to clearly explain concepts.

The most challenging areas were validation workflow and evaluation metrics, especially understanding when and why log-loss should be preferred over accuracy. I also overlooked how improper preprocessing steps, such as scaling before splitting, can cause data leakage.

In addition, I tended to focus on achieving high accuracy rather than considering real-world trade-offs like interpretability and deployment constraints.

Over the next week, I plan to:

- Practice explaining validation strategies clearly
- Review evaluation metrics such as cross-entropy
- Improve my ability to discuss model trade-offs in practical scenarios

## 💬 Talking Points 

1. **Implementation vs Explanation Gap**
   I found that being able to run models is different from being able to explain why certain validation strategies or metrics are used.

2. **Accuracy is Not Always Enough**
   The interview highlighted that accuracy alone may hide prediction confidence issues, and log-loss provides a deeper evaluation.

3. **Hidden Risks of Data Leakage**
   I underestimated how easily leakage can happen through preprocessing steps like scaling before splitting.

4. **Model Choice Requires Trade-offs**
   I initially focused only on performance, but learned that real-world decisions also involve interpretability and deployment constraints.
