Topic 1 — Supervised vs Unsupervised

Interviewer:
Can you briefly explain the difference between supervised and unsupervised learning?

Supervised learning uses labeled data to make predictions, while unsupervised learning finds patterns in data without labels.

Good — that’s correct.

Topic 2 — Regression vs Classification

When would you use linear regression instead of logistic regression?

Linear regression predicts continuous values and logistic regression predicts classification probability.

Right.

Topic 3 — Features vs Target

Can you explain dependent vs independent variables in ML?

Independent variables are inputs and dependent variable is the output.

That’s partially correct — in ML we usually refer to them as features (X) and target (y). In interviews, clarity matters.

Topic 4 — Dataset Splitting

Why do we split into train, validation, and test sets?

Training is used for learning, validation is used for tuning, and test is for final evaluation.

Correct.

Topic 5 — Stratification

What does stratified splitting help with?

It keeps the class distribution balanced across datasets.

Yes — good.

Topic 6 — Data Leakage

What is data leakage?

When test data influences the training process.

Interviewer:
That’s partially correct — can you give an example?

Maybe if we scale data before splitting?

Interviewer:
Exactly — fitting a scaler before splitting leaks information.

Topic 7 — Metrics

Interviewer:
Why might accuracy not be enough?

Because it doesn’t consider prediction confidence.

Good — log-loss helps measure that.

Topic 8 — Cross-Entropy

What does log-loss penalize?

Confident wrong predictions.

Correct.

Topic 9 — Logistic Coefficient

If a coefficient is positive in logistic regression, what does it mean?


That feature increases probability of class 1.

Yes.

Topic 10 — Intercept

What does the intercept represent?

Baseline probability.

Partially — more precisely, baseline log-odds.

Topic 11 — KNN

What happens when k becomes very large?

The model becomes smoother but may underfit.

Interviewer:
Correct.

Topic 12 — Distance Metrics

Can you name two distance metrics used in KNN?

Euclidean and Manhattan.

Right — but remember scaling matters.

Topic 13 — Decision Trees

What does a leaf node represent?

Not quite — it represents the predicted class or value.

Topic 14 — Overfitting

Correct.
